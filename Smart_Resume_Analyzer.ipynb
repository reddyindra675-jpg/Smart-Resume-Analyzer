```python
# Smart Resume Analyzer

resume_skills = input("Enter your resume skills (comma separated): ").lower().split(",")
job_skills = input("Enter required job skills (comma separated): ").lower().split(",")

resume_skills = [skill.strip() for skill in resume_skills]
job_skills = [skill.strip() for skill in job_skills]

matched_skills = []

for skill in job_skills:
    if skill in resume_skills:
        matched_skills.append(skill)

match_percentage = (len(matched_skills) / len(job_skills)) * 100

print("\nMatched Skills:", matched_skills)
print("Match Percentage:", round(match_percentage, 2), "%")

missing_skills = set(job_skills) - set(matched_skills)
print("Missing Skills:", list(missing_skills))
```
